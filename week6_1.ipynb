```python
# ── [개념] sum 으로 개수, 비율 ──────────────────────────────────────────────
# 왜: True=1이라 sum이 곧 결측 개수. 전체 행으로 나누면 비율.
```

```python
print('전체 결측:', df.isna().sum().sum())      # 475
```

```python
missing = df.isna().sum()
```

```python
pd.set_option("display.precision", 1)
```

```python
print((missing / len(df) * 100))       # 컬럼별 비율(%)
```

```python
# ── [개념] 정렬·행별(axis=1)·추출 ───────────────────────────────────────────
```

```python
print(missing.sort_values(ascending=False).head(3))   # 계량종료점·감압시간·최소쿠션
```

```python
row = df.isna().sum(axis=1)                            # 행 방향 결측 수
```

```python
print((row == 0).sum(), (row >= 1).sum())             # 76 174
```

```python
print(missing[missing > 50])    
```
