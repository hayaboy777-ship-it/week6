```python
import matplotlib.pyplot as plt
plt.rcParams["font.family"] = 'Malgun Gothic'
```

```python
PROC = "..\\data\\15_01_사출성형_공정 copy.csv"   # 사출성형 공정 250행 × 22열 (측정시각·불량여부 + 공정변수 20개)
LOG  = "..\\data\\15_사출성형_로그 copy.csv"      # 사출성형 로그 16행 (NaN + 위장 결측 0·-999·999)
```

```python
import pandas as pd
```

```python
df = pd.read_csv(PROC)
```

```python
cnt = df.isna().sum()
```

```python
cnt
```

```python
cnt[cnt > 0].sort_values(ascending=False)
```

```python
cnt[cnt > 0].sort_values(ascending=False).plot
```

```python
cnt[cnt > 0].sort_values(ascending=False).plot.bar()
```

```python
plt.show(cnt[cnt > 0].sort_values(ascending=False).plot.bar())
```
