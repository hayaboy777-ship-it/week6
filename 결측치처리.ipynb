```python
# ── [실습 1] dropna로 행·열 삭제 (사출성형 공정) ────────────────────────────
```

```python
print(df.shape)  # (250, 22)
print(df.dropna().shape)  # (76, 22) 행 삭제
print(df.dropna(axis=1).shape)  # (250, 10) 열 삭제
```

```python
# ── [실습 2] dropna 옵션 조절 (사출성형 공정) ───────────────────────────────
```

```python
print(df.dropna(how="all").shape)  # (250, 22) 거의 유지
print(df.dropna(thresh=20).shape)  # (162, 22)
print(df.dropna(subset=["불량여부"]).shape)  # 불량여부 빈 행만
```

```python
# ── [실습 3] 결측 비율 기준 컬럼 제거 (사출성형 공정) ───────────────────────
#  결측치 비율 0.4 이상인 칼럼(['최대사출속도', '감압시간'])은 제거
```

```python
비율 = df.isna().sum() / len(df)
제거 = 비율[비율 > 0.4].index.tolist()  # ['최대사출속도', '감압시간']
print(제거)
print(df.drop(columns=제거).shape)  # (250, 20)
```

```python
# ── [실습 4] 삭제 손실 비교 (사출성형 공정) ─────────────────────────────────
```

```python
비교 = pd.DataFrame(
    {
        "방식": ["원본", "행삭제", "thresh20"],
        "행": [len(df), len(df.dropna()), len(df.dropna(thresh=20))],
    }
)
비교["손실률"] = ((1 - 비교["행"] / len(df)) * 100).round(1)
비교  # 76(69.6%), 162(35.2%)
```

```python
# ── [실습 5] fillna 평균·중앙값 대체 (사출성형 공정) ────────────────────────
```

```python
print(
    round(df["최대사출압"].mean(), 2), round(df["최대사출압"].median(), 2)
)  # 1241.67 1238.4
평균채움 = df["최대사출압"].fillna(df["최대사출압"].mean())
중앙채움 = df["최대사출압"].fillna(df["최대사출압"].median())
print(평균채움.isna().sum(), 중앙채움.isna().sum())  # 0 0
```
